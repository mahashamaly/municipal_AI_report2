```python
df = pd.read_excel('municipal_complaints_gaza.xlsx',
                   sheet_name='Complaints Data',
                   header=1)
```

```python
# 4. تنظيف الداتا
df_clean = df.copy()
df_clean['Priority'] = df_clean['Priority'].str.strip().str.capitalize()
df_clean['Priority'] = df_clean['Priority'].replace({'Lo': 'Low', 'Meduim': 'Medium'})
df_clean['Status'] = df_clean['Status'].str.strip().str.title()
type_map = {'ELECTRICITY': 'Electricity', 'water': 'Water', 'رود': 'Roads', 'Cnstrctn': 'Construction', 'Wast': 'Waste'}
df_clean['Complaint Type'] = df_clean['Complaint Type'].replace(type_map)
df_clean = df_clean[df_clean['Response Time'] > 0]
df_clean = df_clean.drop_duplicates(subset='ID', keep='first')
df_clean['Name'] = df_clean['Name'].fillna('Unknown')
df_clean['Area'] = df_clean['Area'].fillna('Unknown')
print("✅ تم التنظيف!", df_clean.shape)

```

```python
analysis = f"""
بناءً على تحليل بيانات شكاوى البلدية:
1. إجمالي الشكاوى: {len(df_clean)}
2. أكثر نوع شكاوى: {df_clean['Complaint Type'].value_counts().index[0]} ({df_clean['Complaint Type'].value_counts().values[0]} شكوى)
3. أكثر منطقة شكاوى: {df_clean['Area'].value_counts().index[0]} ({df_clean['Area'].value_counts().values[0]} شكوى)
4. متوسط وقت الاستجابة: {round(df_clean['Response Time'].mean(), 2)} ساعة
5. شكاوى عالية الأولوية: {len(df_clean[df_clean['Priority'] == 'High'])}
"""
print(analysis)
```
